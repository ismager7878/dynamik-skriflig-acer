```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from sympy import pprint
```

## Opg 1

#### 1)

Find the jacobian matrix for the functions

Using the formula:

$$
J = \begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \frac{\partial f_1}{\partial x_2} \\
\frac{\partial f_2}{\partial x_1} & \frac{\partial f_2}{\partial x_2}
\end{bmatrix}
$$
**NOTE** Husk den rigtig rækkefølge af variablerne.

```python
theta, s = sp.symbols('theta s')

f1 = s + 4*sp.cos(theta)
f2 = 3 + 4*sp.sin(theta)

J = sp.Matrix([[sp.diff(f1, s), sp.diff(f1, theta)],
                [sp.diff(f2, s), sp.diff(f2, theta)]])
print("Jacobian matrix:")
sp.pprint(J)
```

#### 2)
To find the singularitets, find where the determinant of the Jacobian is equal to zero.

```python

det_J = J.det()
print("\nDeterminant of the Jacobian:")
pprint(det_J)
sings = sp.solve(det_J, (s, theta))
print("Singularities(Rads):")
pprint(sings)
sings_deg = [(s_val, np.degrees(float(theta_val))) for s_val, theta_val in sings]
print("Singularities(Degree):")
pprint(sings_deg)
```

#### 3) 
Visualize the singularities

```python
plt.figure()

# Select values to get an idea of trajectories
theta_vals = np.array([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
s_vales = [1,2,3,4]



# Visualize the trajectories
x_vals = []
y_vals = []

sing_x_vals = []
sing_y_vals = []

for s_vals in s_vales:
    for theta_val in theta_vals:
        x_vals.append(float(f1.subs({s: s_vals, theta: theta_val}).evalf()))
        y_vals.append(float(f2.subs({s: s_vals, theta: theta_val}).evalf()))

    for s_sing, theta_sing in sings:
        sing_x_vals.append(float(f1.subs({s: s_vals, theta: theta_sing}).evalf()))
        sing_y_vals.append(float(f2.subs({s: s_vals, theta: theta_sing}).evalf()))



plt.plot(x_vals, y_vals, 'o-')
plt.plot(sing_x_vals, sing_y_vals, 'rx', markersize=10, label='Singularities')
plt.legend()
plt.title('Trajectories with Singularities')
plt.xlabel('f1 (x)')
plt.ylabel('f2 (y)')
plt.grid()
plt.axhline(0, color='black',linewidth=0.5, ls='--')
plt.axvline(0, color='black',linewidth=0.5, ls='--')
plt.show()

```
